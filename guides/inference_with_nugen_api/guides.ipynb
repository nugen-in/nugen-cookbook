### **Inference Generate Chat Completions**

This cookbook demonstrates how to run inference using a model that has been aligned and deployed on the Nugen platform. It provides step-by-step instructions for configuring the API client, setting up authentication, specifying the deployed model ID, sending inference requests, and processing the generated responses.

Follow the examples in the order in which they are presented. Before running the code, make sure that your model has been successfully aligned and deployed, and replace all placeholder values with your own API key and deployed model ID.

**Import Required Libraries**

```python
import os 
import requests
import pandas as pd
import json
from dotenv import load_dotenv
load_dotenv()
```

**Set up the Nugen API Client**

**Configure the API Client**

Set your Nugen API key using an environment variable:

```python
API_KEY = os.getenv("NUGEN_API_KEY")
```

Replace your-nugen-api-key with your actual API key before running the script. This key is required to authenticate your requests to the Nugen API.

**Recommendation**

For a better user experience, use "stream": True instead of "stream": False whenever possible. Streaming returns the model's output incrementally as it is generated, which reduces the perceived response time and allows results to be displayed in real time.

### **Step 2**

**Run Inference**

After configuring the API key and the model ID, run the script to send a request to the Nugen Inference API. The API processes the input and returns the generated response.

After completing the alignment process, you must deploy the model before running inference. Once the deployment is complete, use the deployed model ID when making inference requests.

The following model ID is provided only as an example. Replace it with the model ID of your own deployed and aligned model.

```python
MODEL_ID = "model_test-llama-v3p2-3b-reasoning-aligned_alignment_01234*********"
```

```python
inference_url ="https://api.nugen.in/api/v3/inference/chat/completions"
```

```python
payload = {
    "model": MODEL_ID,
    "messages": [
        {
            "role": "user",
            "content": "Tell me about India."
        }
    ],
    "max_tokens": 100,
    "temperature": 0.1,
    "stream": True
}

headers = {
    "accept": "application/json",
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}

response = requests.post(
    inference_url,
    json=payload,
    headers=headers,
    stream=True
)

for line in response.iter_lines():
    if line:
        print(line.decode("utf-8"))
```
